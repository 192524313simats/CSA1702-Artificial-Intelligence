```python
# Monkey and Banana Problem

monkey = "door"
box = "window"
banana = "ceiling"

print("Initial State:")
print("Monkey is at", monkey)
print("Box is at", box)
print("Banana is at", banana)

print("\n1. Monkey moves to the box.")
monkey = box

print("2. Monkey pushes the box below the banana.")
box = "under banana"
monkey = "under banana"

print("3. Monkey climbs the box.")
monkey_height = "high"

if monkey == "under banana" and monkey_height == "high":
    print("4. Monkey gets the banana!")
    print("\nGoal Achieved!")
```
