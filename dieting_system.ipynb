```python
# Dieting System based on Disease

diet = {
    "diabetes": [
        "Eat vegetables and whole grains",
        "Avoid sugary foods"
    ],
    "obesity": [
        "Eat fruits and vegetables",
        "Avoid oily and junk foods"
    ],
    "anemia": [
        "Eat iron-rich foods",
        "Eat green leafy vegetables"
    ]
}

disease = input("Enter disease: ").lower()

if disease in diet:
    print("\nDiet suggestions:")
    for food in diet[disease]:
        print("-", food)
else:
    print("Disease not found.")
```
