```python
# Backward Chaining Example

# Known facts
facts = {"rain"}

# Rule: rain requires carry umbrella
rules = {
    "rain": "carry umbrella"
}

# Goal
goal = "carry umbrella"

print("Goal:", goal)

if goal in rules:
    required_fact = rules[goal]
    print("Need to prove:", required_fact)

    if required_fact in facts:
        print("Fact found:", required_fact)
        print("Goal Proved")
    else:
        print("Goal Cannot Be Proved")
else:
    print("No rule available")
```

```python
# Forward Chaining Example

# Initial fact
facts = {"rain"}

# Rule: If Study then Pass
if "rain" in facts:
    facts.add("carry umbrella")

# Print inferred facts
print("Knowledge Base:")
for fact in facts:
    print(fact)
```

```python
# Knowledge Representation using Python

# --------------------------
# Facts
# --------------------------
facts = {"Cat(Tom)"}

# --------------------------
# Rules
# Key = Conclusion
# Value = Required Fact
# --------------------------
rules = {
    "Mammal(Tom)": "Cat(Tom)",
    "Animal(Tom)": "Mammal(Tom)"
}

print("Initial Facts:")
for fact in facts:
    print(fact)

print("\nApplying Rules...")

# Keep applying rules until no new facts are added
changed = True

while changed:
    changed = False

    for conclusion, premise in rules.items():

        if premise in facts and conclusion not in facts:
            facts.add(conclusion)
            print(f"Derived: {conclusion}")
            changed = True

print("\nFinal Knowledge Base:")
for fact in facts:
    print(fact)
```
