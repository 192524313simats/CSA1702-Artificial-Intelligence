```python
# Knowledge Base
attendance = 80
marks = 65

# Reasoning
if attendance >= 75 and marks >= 50:
    print("Knowledge: Attendance and marks satisfy the rules.")
    print("Reasoning: Student is eligible for the exam.")
else:
    print("Knowledge: Requirements are not met.")
    print("Reasoning: Student is not eligible.")
```

```python
# -------------------------------
# Backward Chaining Example
# -------------------------------

# Facts stored in knowledge base
facts = {
    "Human(Socrates)"
}

# Rules stored as:
# conclusion : premise
rules = {
    "Mortal(Socrates)": "Human(Socrates)"
}

# Recursive backward chaining function
def backward_chain(goal):

    print(f"\nTrying to prove: {goal}")

    # Step 1: Check whether goal is already a fact
    if goal in facts:
        print(f"{goal} is a FACT.")
        return True

    # Step 2: Check if any rule can prove it
    if goal in rules:

        premise = rules[goal]

        print(f"Need to prove: {premise}")

        if backward_chain(premise):
            print(f"Rule applied:")
            print(f"{premise} --> {goal}")
            return True

    print(f"Cannot prove {goal}")
    return False


# Goal
goal = "Mortal(Socrates)"

print("Backward Chaining")

if backward_chain(goal):
    print("\nGoal PROVED")
else:
    print("\nGoal NOT PROVED")
```

```python
# Forward Chaining Example

# Initial fact
facts = {"rains"}

# Rule: If Study then Pass
if "rains" in facts:
    facts.add("carry umbrella")

# Print inferred facts
print("Knowledge Base:")
for fact in facts:
    print(fact)
```
